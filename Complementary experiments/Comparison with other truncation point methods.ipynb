```r
for (j in 1:1)
{
datanamelist=c("Hourly","Daily","Weekly","Monthly","Quarterly","Yearly")
data = Filter(function(l) l$period == datanamelist[j], M4)
datalength = length(data)
h=data[[1]]$h
freq=frequency(data[[1]]$x)

set.seed(100)
index = sample(2,length(data),replace = TRUE,prob=c(0.7,0.3))
trainindex=c(1:length(data))[index==1]
testindex=c(1:length(data))[index==2]
mape_ets <- c()
mape_theta <- c()
mase_ets <- c()
mase_theta <- c()
for(i in testindex){
    hdata = data[[i]]$x
    real = data[[i]]$xx
    cpm.res = detectChangePoint(hdata, cpmType = "Kolmogorov-Smirnov")
    hdata <- ts(hdata[cpm.res$changePoint:length(hdata)],frequency=freq, end=end(data[[i]]$x))
    
    ets_model <- ets(hdata)
    forecast_ets <- forecast(ets_model, h = h)
    forecast_theta <- thetaf(hdata, h = h)
    
    res_ets = accuracy(forecast_ets,real)
    res_theta = accuracy(forecast_theta,real)
    mape_ets = c(mape_ets, res_ets[2,5])
    mape_theta = c(mape_theta, res_theta[2,5])
    mase_ets = c(mase_ets, res_ets[2,6])
    mase_theta = c(mase_theta, res_theta[2,6])
}    
}
```

```r
mean(mape_theta)
mean(mase_theta,na.rm = TRUE)
```
