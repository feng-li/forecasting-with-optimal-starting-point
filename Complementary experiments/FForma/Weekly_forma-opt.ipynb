```r
pre_forecast_fforma=function(data,freq,h,m,n,meta_model)
{
    datapre=data
    model=ets(data[[1]]$x)
    fore_l=forecast(model,h=h)
    data_length=length(data)
    pred = array(0,dim = c(data_length,m,n,6))
    predh = array(0,dim = c(data_length,m,n,h))
    for(k in 1:data_length){
        start_time = Sys.time()
        y = data[[k]]$x
        y_pred = data[[k]]$xx
        y_l=length(y)
        loc = 1:length(y)
        loc_m = as.integer(loc*m/length(y))
        filt_d = data.frame(y,loc_m)
        filt_0=filter(filt_d ,loc_m==0)
        m_l=count(filt_0)
        n_l=m_l/n
        m_l=as.integer(m_l)
        n_l=as.numeric(n_l)
        for(i in 0:(m-1)){
            for(j in 0:(n-1)){
                Y = ts(y[round((i*m_l+j*n_l)+1):y_l], end = end(y), frequency=freq)
                datapre[[k]]$x=Y
                datapre[k] <- calc_forecasts(datapre[k], forec_methods(), n.cores=1)
                datapre[k] <- THA_features(datapre[k], n.cores=1)
                test_data <- create_feat_classif_problem(datapre[k])
                preds <- predict_selection_ensemble(meta_model, test_data$data)
                datapre[k] <- ensemble_forecast(preds, datapre[k])
                #print(datapre[[k]]$y_hat)
                fore_l$x=Y
                fore_l$mean=ts(datapre[[k]]$y_hat,start=start(datapre[[k]]$xx),frequency=freq)
                predh[k,(i+1),(j+1),] =datapre[[k]]$y_hat 
                pred[k,(i+1),(j+1),] = accuracy(fore_l,datapre[[k]]$xx)[2,1:6]
            }
        }
}
    return(list(pred, predh))
}
```

```r
predres=function(data)
    {
    data_length=length(data)
    pred_res=matrix(0,data_length,6)
    model=nnetar(data[[1]]$x)
    h=data[[1]]$h
    fore_l=forecast(model,h=h)
    freq=frequency(data[[1]]$x)
    for (i in 1:data_length){
    fore_l$x=data[[i]]$x
    real=data[[i]]$xx
    fore_l$mean=ts(data[[i]]$y_hat,start=start(real),frequency=freq)
    res=accuracy(fore_l,real)
    pred_res[i,]=res[2,1:6]
    }
    return(pred_res)
    }
```

```r
pred_res=predres(M4_test)
```

```r
mean1=function(x)
    {
# 使用is.finite函数检查哪些值是有限的
finite_values <- x[is.finite(x)]
# 计算有限值的平均值
d=mean(finite_values, na.rm = TRUE)
return(d)
}
```
